```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))

print("Pick an operation: 1. Addition  2. Subtraction  3. Multiplication  4. Division")

def Addition():
  sum = num1 + num2
  return sum

def Subtraction():
  difference = num1 - num2
  return difference

def Multiplication():
  product = num1 * num2
  return product

def Division():
  quotient = num1 / num2
  return quotient

while True:
  operation_choice = int(input("Enter operation number: "))

  if operation_choice > 4 or operation_choice < 1:
    print("Invalid operation, please try again.")

  else:
    break

if operation_choice == 1:
  print(f"The sum is: {Addition()}")

if operation_choice == 2:
  print(f"The difference is: {Subtraction()}")

if operation_choice == 3:
  print(f"The product is: {Multiplication()}")

if operation_choice == 4:
  print(f"The quotient is: {Division()}")
```
