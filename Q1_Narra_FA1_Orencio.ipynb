```python
import math

#Make the variables set up the coordinates.
x1 = int(input("enter x1: "))
y1 = int(input("enter y1: "))
x2 = int(input("enter x2: "))
y2 = int(input("enter y2: "))

#Make the formula calculate the distance for both coordinates.
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))
#Print the final answer.
rounded = round(distance, 2)
print("the distance is: ", rounded)

#1. It helped simplify my math code because it uses a math library which adds math functions instead of maually writing them with code.
#2. The functions that were easier to use with the math library is math.sqrt and math.pow.
#3. Without the sqrt() or pow(), it would be so much more time consuming and harder because those math funcions act like a shortcut in coding.

```
